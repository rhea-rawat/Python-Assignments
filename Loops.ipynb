```python
data=[18,29,44,-3,76,-6,92,-1]
posCount=0
negCount=0
i=0
while(i<len(data)):
    if data[i]>0:
        posCount+=1
    else:
        negCount+=1
    i+=1
print(posCount)
print(negCount)
```
