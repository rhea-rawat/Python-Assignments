```python
numbers = [14, -7, 25, -4, 36, -9]
result=[ex if ex>0 else 0 for ex in numbers]
print(result)
```

```python
transactions = [6500, -2500, 4200, -1500, 8000, -700]
deposits=[ex for ex in transactions if ex>0]
print(deposits)
```

```python
prices = [650, 1400, 900, 2400, 1800]
result=[ex*0.90 if ex>=1000 else ex*0.95 for ex in prices]
print(result)
```
